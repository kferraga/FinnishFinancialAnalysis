# Candidate Fundraising Analysis

```python
# PREPARING DATA FOR MODEL AND VISUALIZATION
# STEP 1 - Importing Sheets
import pandas as pd
pd.set_option('display.max_columns', None)

kuntavaalit_ehdokas_sarakkeet = list(pd.read_csv("files/Results_title_rows_EN_ehdokas.csv")) # Saved as a list to apply to any candidate voting information, because header wasn't included in the data from the Ministry of Justice.

years_data = {
    2011: {
        'fundraising': "files/finances/Liite_1_Eduskuntavaalit_2011_RAHOITUSRIVIT_VI.csv"
    },
    2015: {
        'fundraising': "files/finances/Liite_3_Eduskuntavaalit_2015_RAHOITUSRIVIT_VI.csv"    
    },
    2019: {
        'fundraising': "files/finances/Liite_5_Eduskuntavaalit_2019_RAHOITUSRIVIT_VI.csv"
    },
    2023: {
        'fundraising': "files/finances/Liite_7_Eduskuntavaalit_2023_RAHOITUSRIVIT_VI.csv"
    }
}

all_years_data = []
for year, paths in years_data.items():
    print(f"Processing {year}...")
    year_df = pd.read_csv(paths['fundraising'], sep=";", quotechar="'", on_bad_lines='warn')
    year_df["Vuosi"] = year
    year_df["Tuen maara"] = year_df["Tuen maara"].replace({',':'.', " ":""}, regex=True).astype(float)
    all_years_data.append(year_df)

kuntavaalit_yhteensä_all = pd.concat(all_years_data, ignore_index=True)
print(f"In total, there were {len(all_years_data[0])} candidates recorded from 2012; {len(all_years_data[1])} from 2017; {len(all_years_data[2])} from 2021; and {len(all_years_data[3])} from 2025.")
display(kuntavaalit_yhteensä_all)
```

```python
kuntavaalit_yhteensä_all.to_csv("files/outputs/eduskuntavaalit_financial_line_items.csv", index=False)

kuntavaalit_yhteensä_2023 = kuntavaalit_yhteensä_all[kuntavaalit_yhteensä_all["Vuosi"]==2023]
kuntavaalit_yhteensä_all.groupby(["Tukija", "Tuen saajan puolue"])[["Tuen maara"]].mean()
```
